```python
def pattern22(n):
    for i in range(2 * n - 1):
        for j in range(2 * n - 1):
            value = n - min(i, j, 2 * n - 2 - i, 2 * n - 2 - j)
            print(value, end=" ")
        print()


pattern22(5)
```

```python
def pattern21(n):
    for row in range(n):
        for col in range(n):
            if row == 0 or row == n-1 or col == 0 or col == n-1:
                print("*", end="")
            else:
                print(" ", end="")
        print()

pattern21(5)
```

```python
def pattern20(n):
    for row1 in range(n):
        print("*"*(row1+1), end="")
        print(" "*2*(n-(row1+1)), end="")
        print("*"*(row1+1), end="")
        print()

    for row2 in range(n-1):
        print("*"*(n-(row2+1)), end="")
        print(" "*2*(row2+1), end="")
        print("*"*(n-(row2+1)), end="")
        print()

pattern20(5)
```

```python
def pattern19(n):
    for row1 in range(n):
        print("*"*(n-row1), end="")
        print(" "*2*row1, end="")
        print("*"*(n-row1), end="")
        print()

    for row2 in range(n):
        print("*"*(row2+1), end="")
        print(" "*2*(n-row2-1), end="")
        print("*"*(row2+1), end="")
        print()

pattern19(5)
```

```python
def pattern18(n):
    num = n
    for row in range(n):
        for col in range(row+1):
            print(chr(65+num+col-1), end=" ")
        num -= 1
        print()

pattern18(5)
```

```python
def pattern17(n):
    for row in range(n):
        print(" "*(n-row-1), end="")

        for col1 in range(row+1):
            print(chr(65 + col1), end="")

        for col2 in range(row, 0, -1):
            print(chr(65 + col2-1), end="")

        print()

pattern17(5)
```

```python
def pattern13(n):
    num = 1
    for row in range(n):
        for col in range(row + 1):
            print(num, end=" ")
            num += 1
        print()

pattern13(5)
```

```python
def pattern12(n):
    for row in range(n):
        for col1 in range(row + 1):
            print(col1+1, end="")

        for _ in range(2*(n-(row+1))):
            print(" ", end="")

        for col3 in range(row+1, 0, -1):
            print(col3, end="")
        print()

pattern12(5)
```

```python
def pattern11(n):
    for row in range(n):
        for col in range(row + 1):
            print((n-col)%2, end="")
        print()

pattern11(5)
```

```python
def pattern10(n):
    for row in range(n):
        print("*"*(row + 1))

    for row in range(n-1):
        print("*"*(n-(row+1)))

pattern10(5)
```

```python
def pattern9(n):
    for row in range(n):
        for col in range(n-row-1):
            print(" ", end="")

        for col in range(2*row+1):
            print("*", end="")
        print()

    for row in range(n):
        for col in range(row):
            print(" ", end="")

        for col in range(2*(n-row)-1):
            print("*", end="")
        print()

pattern9(5)
```

```python
def pattern8(n):
    for row in range(n):
        for col in range(row):
            print(" ", end="")

        for col in range(2*(n-row)-1):
            print("*", end="")
        print()

pattern8(5)
```
