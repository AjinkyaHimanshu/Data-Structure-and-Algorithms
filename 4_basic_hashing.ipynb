### **Sum of Highest and Lowest Frequency**

```python
def sumHighestAndLowestFrequency(nums):
    occDict = {}

    for num in nums:
        occDict[num] = occDict.get(num, 0) + 1
    maxOcc = max(occDict.values())
    minOcc = min(occDict.values())
    return maxOcc + minOcc

sumHighestAndLowestFrequency([1, 2, 2, 3, 3, 3])
```

### **Second Highest Occurring Element**

```python
def secondMostFrequentElement(nums):
    occDict = {}

    for num in nums:
        occDict[num] = occDict.get(num, 0) + 1
    maxOcc = max(occDict.values())
    secondMaxOcc = max(value for value in occDict.values() if value < maxOcc)
    return next(key for key, value in occDict.items() if value == secondMaxOcc)

secondMostFrequentElement([1, 2, 2, 3, 3, 3])
```

### **Highest Occurring Element in an Array**

```python
def mostFrequentElement(nums):
    occDict = {}

    for num in nums:
        occDict[num] = occDict.get(num, 0) + 1
    maxOcc = max(occDict.values())
    return next(key for key, value in occDict.items() if value == maxOcc)

mostFrequentElement([1, 2, 2, 3, 3, 3])
```
