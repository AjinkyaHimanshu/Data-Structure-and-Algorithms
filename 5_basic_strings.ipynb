### **Sort Characters by Frequency**

```python
def frequencySort(s):
    freq = {}
    for char in sorted(s):
        freq[char] = freq.get(char, 0) + 1
    return list(freq.keys())

frequencySort("raaaajj")
```

### **Valid Anagram**

```python
def anagramStrings(s, t):
    if len(s) != len(t):
        return False
    return sorted(s) == sorted(t)

anagramStrings("listen", "silent")
```

### **Rotate String**

```python
def rotateString(s, goal):
    if len(s) != len(goal):
        return False
    return goal in (s + s)

rotateString("abcde", "adeac")
```

### **Isomorphic Strings** <span style="color: red;">*</span>

```python
# Given two strings s and t, determine if they are isomorphic. 
# Two strings s and t are isomorphic if the characters in s can be replaced to get t.

# All occurrences of a character must be replaced with another character while preserving the order of characters. 
# No two characters may map to the same character, but a character may map to itself.
def isomorphicString(s : str, t : str) -> bool:
    if len(s) != len(t):
        return False

    s_to_t = {}
    t_to_s = {}

    for a,b in zip(s,t):
        if a in s_to_t and s_to_t[a] != b:
            return False

        if b in t_to_s and t_to_s[b] != a:
            return False

        s_to_t[a] = b
        t_to_s[b] = a

    return True
    

isomorphicString("egg", "add")
```

### **Longest Common Prefix** <span style="color: red;">*</span>

```python
def longestCommonPrefix(strs):
    '''
    Find the longest common prefix among an array of strings.
    '''
    if not strs:
        return ""

    for i, char in enumerate(strs[0]):
        for string in strs[1:]:
            if i >= len(string) or string[i] != char:
                return strs[0][:i]

    return strs[0]


# longestCommonPrefix(["flowers" , "flow" , "fly", "flight" ])
longestCommonPrefix(["dog" , "cat" , "animal", "monkey" ])
```

### **Palindrome Check**

```python
def palindromeCheck(s):
    return s[::-1]==s

palindromeCheck('hannah')
```

### **Reverse a String II**

```python
def reverseString(s):
    return s[::-1]

reverseString(["h", "e" ,"l" ,"l" ,"o"])
```
