### **Divisors of a Number**

```python
def divisors(n):
    divs = []

    for num in range(1, int(n**(0.5))+1):
        if n%num==0:
            divs += [n//num, num]
    return sorted(divs)

divisors(6)
```

### **LCM of two numbers**

```python
def LCM(n1, n2):
    return (n1 * n2) // GCD(n1, n2)

LCM(45,10)
```

### **GCD of Two Numbers**

```python
def GCD(n1, n2):
    if n2 > n1:
        n1, n2 = n2, n1
    # print(f"n1:{n1}, n2:{n2}")

    while n1>1:
        if n1%n2==0:
            return n2
        n1, n2 = n2, n1%n2
        # print(f"n1:{n1}, n2:{n2}")
    return 1

GCD(10,999)     
```

### **Check for Prime Number**

```python
def isPrime(n):
    root = int(n**(0.5))

    for num in range(root, 1, -1):
        if n%num==0:
            return False
    return True

isPrime(43)
```

### **Check for Perfect Number**

```python
def isPerfect(n):
    half = n//2
    divisorSum = 0

    for num in range(1, half+1):
        if n%num==0:
            divisorSum += num
    return divisorSum == n

isPerfect(28)
```

### **Check if the Number is Armstrong**

```python
def isArmstrong(n):
    sum = 0
    actualN = n
    while n > 0:
        digit = n % 10
        sum += digit**3
        n //= 10
        
    return sum==actualN

isArmstrong(153)
```

### **Factorial of a given number**

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

factorial(5)
```

### **Return the Largest Digit in a Number**

```python
def largetstDigit(n):
    largest = 0
    while n > 0:
        digit = n % 10
        if digit > largest:
            largest = digit
        n //= 10
    return largest

print(largetstDigit(25))
```

### **Palindrome Number**

```python
def isPalindrome(n):
    return str(n) == str(n)[::-1]

print(isPalindrome(123)) 
```

### **Reverse a number**

```python
def reverseNumber(n):
    reversed_num = 0
    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n //= 10
    return reversed_num

print(reverseNumber(234))  # Output: 432
```

### **Count number of odd digits in a number**

```python
def countOddDigit(n):
    count = 0
    for digit in str(n):
        if int(digit) % 2 == 1:
            count += 1
    return count

print(countOddDigit(25))  # Output: 1
```

#### **Count all Digits of a Number**

```python
def countDigit(n):
    return len(str(n))

print(countDigit(234))  # Output: 3
```
